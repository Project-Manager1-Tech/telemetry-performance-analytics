```python
import pandas as pd

df = pd.read_excel("../data/raw/IoT_Telemetry_Dataset_100k.xlsx")
df['timestamp'] = pd.to_datetime(df['timestamp'])
df['date'] = df['timestamp'].dt.date
df['hour'] = df['timestamp'].dt.hour
df['month'] = df['timestamp'].dt.month
df.head()
df['risk_flag'] = (
    (df['cpu_usage'] > 85)
    |
    (df['temperature_c'] > 70)
    |
    (df['network_latency_ms'] > 120)
)
df['error_severity'] = pd.cut(
    df['error_count'],
    bins=[-1,0,2,5,100],
    labels=[
        'No Error',
        'Low',
        'Medium',
        'High'
    ]
)
df['cpu_category'] = pd.cut(
    df['cpu_usage'],
    bins=[0,40,70,85,100],
    labels=[
        'Low',
        'Normal',
        'High',
        'Critical'
    ]
)
df['latency_category'] = pd.cut(
    df['network_latency_ms'],
    bins=[0,50,100,150,1000],
    labels=[
        'Excellent',
        'Good',
        'Poor',
        'Critical'
    ]
)
df['device_health_score'] = (
    (100 - df['cpu_usage'])*0.3 +
    (100 - df['memory_usage'])*0.2 +
    (100 - df['network_latency_ms']/2)*0.2 +
    (df['uptime_pct'])*0.3
)
df.to_csv("../data/processed/telemetry_clean.csv",
    index=False)
```
